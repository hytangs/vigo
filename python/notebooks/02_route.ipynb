# Route

Run one Route or reuse Route Query objects as a batch.


```python
import vigo

with vigo.open("./city") as city:
    query = vigo.Route(
        "A",
        "B",
        depart_at="08:00",
        service_date="2026-09-04",
    )
    result = city.run(query)
    batch = city.run([query, query])

print(result.status, result.duration_minutes)
print(len(batch))
```
