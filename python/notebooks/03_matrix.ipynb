# Matrix

A Matrix accepts one or many origins and destinations.


```python
import vigo

with vigo.open("./city") as city:
    result = city.matrix(
        {"home": "A"},
        {"school": "B", "hospital": "C"},
        depart_at="08:00",
        service_date="2026-09-04",
    )

for row in result.rows:
    print(row["originId"], row["destinationId"], row["durationMinutes"])
```
