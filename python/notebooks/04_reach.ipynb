# Reach

Map where the represented network can travel within stated time limits.


```python
import vigo

with vigo.open("./city") as city:
    result = city.reach(
        [-71.062, 42.356],
        depart_at="08:00",
        service_date="2026-09-04",
        cutoffs_minutes=[15, 30, 45, 60],
    )

result.export("reach.geojson")
```
